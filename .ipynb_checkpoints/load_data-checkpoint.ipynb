```python
import json
```

```python
#funtion for load the data

def load_data(filename):
    with open(filename) as f:
        data=json.load(f)
    return data
```

```python
data=load_data("data.json")
```

```python
#function to display the user and their connection 
def display_users(data):
    print("users and their connections: ")
    for user in data['users']:
        
        print(f"ID:{user['id']}- {user['name']}has friendship with :{user['friends']}- Liked Pages: {user['liked_pages']}")
    print(f"\nlist of pages: ")
    for page in data['pages']:
        # id=user['friends']
        print(f"{page['id']} : {page['name']}")
```

```python
display_users(data)
```
